```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Load Seaborn built-in datasets
```

```python
tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
iris = sns.load_dataset("iris")
```

```python
# Create a scatter plot for the tips dataset
```

```python
# scatterplot
sns.scatterplot(data=tips, x="total_bill", y="tip" , hue="day")
plt.savefig("scatter_day.png", dpi=300)
```

```python
#  Create a line plot for the flights dataset
```

```python
# line graph
from matplotlib.typing import LineStyleType
sns.lineplot(data=flights , x="year" , y="passengers" , hue="month" , linestyle="dashed")
plt.savefig("line-month-hue.png" , dpi=300)
```

```python
sns.lineplot(data=flights , x="year" , y="passengers")
plt.savefig("line-month.png" , dpi=300)
```

```python
sns.lineplot(data=flights , x="month" , y="passengers" , hue="year" , linestyle="dashed")
plt.savefig("line-year.png" , dpi=300)
```

```python
#  Create a bar plot for the tips dataset
```

```python
# bargraph
sns.barplot(data=tips , x="day" , y="tip" )
plt.savefig("bar.png" , dpi=300)
```

```python
#  Create a histplot for the tips dataset
```

```python
# histogram
sns.histplot(data=tips , x="total_bill")
#for histogram there is only one quantity
plt.savefig("histogram.png")
```

```python
#  Create a boxplot for the tips dataset
```

```python
sns.boxplot(data=tips, x="day", y="total_bill")
plt.savefig("box.png")
```

```python
sns.countplot(data=tips , x="day")
```

```python
sns.regplot(data=tips , x="total_bill" , y="tip")
plt.savefig("regplot")
```

```python
grid = flights.pivot(index="month" , columns="year" , values="passengers")
```

```python
sns.heatmap(data=grid , cmap="Blues")
plt.savefig("heatmap.png")
```
