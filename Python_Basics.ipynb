```python
# 6. Loops
```

```python
order_amounts = [123 , 1250 , 517 , 980 , 1499 , 340]
print(len(order_amounts))
total_sale = 0
for one_order in order_amounts:
   total_sale = total_sale + one_order
print("total sale is" , total_sale)
```

```python
#how many orders are eligible for free delivery
free_delivery_count = 0
for order_amount in order_amounts :
  if order_amount > 1000 :
    free_delivery_count = free_delivery_count + 1
  print(free_delivery_count)
```
